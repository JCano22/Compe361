**GCD function** example with `return`:

```python
def GCD(p, q):
    while p != q:
        if p > q:
            p = p - q
        else:
            q = q - p
    return p

u = 75
z = 45

r = GCD(u, z)

print(r)
print(u)
print(z)
```

**LCM function** example:

-Least common multiple is the *smallest* number that can be divided by both numbers.

- We use the double slash: `//` for integer division

```python
def GCD(p, q):
    while p != q:
        if p > q:
            p = p - q
        else:
            q = q - p
    return p


def LCM(a, b):
    return a * b // GCD(a,b)

u = 75
z = 45

r = LCM(u,z)

print(r)
print(u)
print(z)
```

**Example of function** parameters vs arguments:

```python
def Exchange(p, q):
    r = p
    p = q
    q = r

u = 75
z = 89

Exchange(u, z)
print(u)
print(z)

# when the function is called, a copy of the variables is sent to the function.  Even though the variables are pointers to the address, they don't send the address to the function.  Therefore the arguments aren't modified in the function.
```

For the above example:

- In Python, variables are names bound to objects, not memory locations or pointers like in C.

- When `Exchange(u, z)` is called:

    - `p` is bound to the same object as `u`

    - `q` is bound to the same object as `z`

- The function receives **copies of the references**, not the variable names `u` and `z` themselves.

- Inside the function, statements like `p = q` and `q = r`:

    - Rebind the local names `p` and `q`

    - Do not affect the bindings of `u` and `z` in the caller

- Integers are **immutable**, so their values cannot be modified in place.

- As a result, swapping `p` and `q` only changes local variables inside the function.

- Therefore, `u` and `z` remain unchanged after the function call.

- You can make another class be a 'child' class of another. 
    - Here `Cuboid` is a child class of `Rectangle`.
    - It inherits the methods from the parent class
    - You can call the parent methods for the `Cuboid` objects using `super()`

Example:

```python
# parent class
class Rectangle:
    length = None
    width = None

    def __init__(self, l, w):
        self.length = l
        self.width = w
    
    def area(self):
        return (self.length * self.width)
    
    def perimeter(self):
        return (2 * self.length + 2 * self.width)
    
# child class
class Cuboid(Rectangle):
    pass

r = Rectangle(2, 4)
a = r.area()
p = r.perimeter()

print(r.length)
print(a)
print(p)

c = Cuboid(9, 3)
b = c.area()
print(f'Cuboid area: {b}')
```

- You can add its own methods as well

- In code below, we overwrite the `__str__` method inside the parent class.  We call to print cuboid object, and it calls parent class `__str__`. But it calls child class method

```python
# parent class
class Rectangle:
    length = None
    width = None

    def __init__(self, l, w):
        self.length = l
        self.width = w
    
    def area(self):
        return (self.length * self.width)
    
    def perimeter(self):
        return (2 * self.length + 2 * self.width)
    
    #overwriting __str__ method
    def __str__(self):
        a = self.area()
        return "The area is: " + str(a)
    
# child class
class Cuboid(Rectangle):
    height = None

    def __init__(self, l, w, h):
        self.height = h
        #you can call the parent's constructor for the remaining values (l, w) using `super` followed by the method you wish to call
        super().__init__(l, w)
    
    def area(self):
        return 2 * super().area() + 2 * (self.length + self.width) * self.height


r = Rectangle(2, 4)
a = r.area()
p = r.perimeter()

print(r.length)
print(a)
print(p)

c = Cuboid(9, 5, 2)
b = c.area()

print(f'Cuboid area: {b}')

print(c) #currently prints address as there is no __str__ method
```

```python
class Piece:
    x = None
    y = None
    def __init__(self, x0, y0):
        self.x = x0
        self.y = y0
    def __str__(self):
        return str(self.x) + ":" + str(self.y)
    def move(self, x1, y1):
        if self.verify(x1,y1):
            self.x = x1
            self.y = y1


class Rook(Piece):
    def verify(self, x1, y1):
        return self.x == x1 or self.y == y1
    
class Bishop(Piece):
    info = 'abcd'
    def verify(self, x1, y1):
        return abs(x1-self.x) == abs(y1-self.y)

class Queen(Rook, Bishop):
    def verify(self,x1,y1):
        return super().verify(x1,y1) or abs(x1-self.x) == abs(y1-self.y)

p = Piece(3,2)
#p.move(4,5) -- incorrect invocation, only avaialble to Rook obj

r = Rook(1,8)
r.move(3,8)
print(r)
r.move(4,5)
print(r)
print("=============")
b = Bishop(3,1)
b.move(5,3)
print(b)
b.move(4,7)
print(b)
print("=============")
q = Queen(5,8)
q.move(1,8)
print(q)
q.move(2,7)
print(q)
print(q.info)
print("=============")
```

```python
class MyTime:
    ticks = 0
    def __init__(self, h, m, s):
        self.ticks = (h * 60 + m) * 60 + s
    
    def gethour(self):
        return self.ticks // 3600
    def getminute(self):
        return self.ticks // 60 % 60
    def getsecond(self):
        return self.ticks % 60
    def sethour(self, h):
        print("I'm set hour. value of h is: " + str(h))
        self.ticks = self.ticks % 3600 + h * 3600
    # property
    hour = property(fget = gethour, fset = sethour)
    minute = property(fget = getminute)
    second = property(fget = getsecond)

u = MyTime(9, 5, 37)
print(u.ticks)
print(u.hour)
print(u.minute)
print(u.second)
u.hour = 19
print(u.hour)
```

### Abstract Classes
- An *Abstract Class* is a class from which you cannot create an object.

- When to use *Abstract Classes*:


- *Abstract Base Classes* cannot be instantiated themselves but can be extended by subclasses.

 

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def volume(self):
        pass

class Rectangle(Shape):
    length = None
    width = None
    def __init__(self, l, w):
        if l <= 0 or w <= 0 or l < w:
            raise Exception("Incorrect arguments")
        self.length = l
        self.width = w
    
    def area(self):
        return self.length * self.width
    def volume(self):
        return 0
    pass

r = Rectangle(20, 3)
a = r.area()
print(a)
```

- Methods in class must override Abstract Class methods in order to use it.

Another examples of *Abstract Class* use:

```python
from abc import ABC, abstractmethod

class Prism(ABC):
    height = None
    def __init__(self, h):
        self.height = h

    @abstractmethod
    def basearea(self):
        pass
    
    def volume(self):
        return self.height * self.basearea()

class Cylinder(Prism): #inherited from Prism, but can now calculate basearea
    radius = None
    def __init__(self, h, r):
        super().__init__(h) # calling parent class constructor w/ value of height
        self.radius = r
    
    def basearea(self):
        return pi * pow(self.radius, 2)

c = Cylinder(7, 3)
v = c.volume()
print(v)
```

### ***Decorators***
- Decorators let you add extra behavior to a function, without changing the function's code.

- A decorator is a function that takes another function as input and returns a new function.

- A decorator in Python is a way to modify or extend the behavior of a function (or class) without changing its actual code.

```python
def info(func):
    def wrapper(z):
        print("before")
        if z < 0:
            print("incorrect argument")
        else:
            func(z)
        print("after")
    return wrapper

@info
def echo(u):
    print(u)

echo(-7)
```

```python
class Foo:
    x = 1000
    z = 2000
    def mydecorator(func):
        def wrapper(self):
            print(self.x)
            func(self)
            print(self.z)
        return wrapper

    @mydecorator
    def echo(self):
        print(self.x + self.z)

f = Foo()
f.echo()
    
```

### ***Descriptors***

- A descriptor is an object that defines how attribute access is handled in another class.

- Descriptors allow you to customize attribute behavior.

- Instead of storing and retrieving attributes normally, Python can run custom logic whenever an attribute is accessed or modified.

```python
class WeekDayDesc:
    def __set__(self, obj, value):
        if value == "Monday" or value == "Tuesday":
            obj._weekdayname = value
        else:
            raise Exception("Incorrect weekday")
    def __get__(self, obj, type=None):
        return obj._weekdayname

class Lecture:
    coursename = None
    _weekdayname = None
    weekday = WeekDayDesc()

u = Lecture()
u.coursename = "Mathematical analysis"
u.weekday = "Monday"

print(u.weekday)
```

### *Iterators*

- An iterator is an object that contains a countable number of values.

- An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.

#### *Context Manager Protocol*

```python
class Foo:
    def __init__(self, x):
        self.x = x
    def __str__(self):
        return str(self.x * self.x)
    def __enter__(self):
        print("enter is called")
        return self
    def __exit__(self, *args):
        print("exit is called")
        print(args)
        return True
    
with Foo(3) as f:
    print(f)
print("good bye")
```

#### ****args*** / *****kwargs***

### *iterators* and *iterables*

- The *iterable* protocol is used by types where it is possible to process their contents one at a time in turn. An iterable is something that will supply an iterator that can be used to perform this processing. As such it is not the iterator itself; but the provider of the iterator.

- An *iterator* is an object that will return a sequence of values. Iterators may be finite in length or infinite (although many container-oriented iterators provide a fixed set of values).

- To be an *iterable* type; it is necessary to implement the `__iter__()` method (which is the only method in the iterable protocol). This method must supply a reference to the iterator object. This reference could be to the data type itself or it could be to another type that implements the iterator protocol.

- The *iterator* protocol specifies the `__next__()` method. This method is expected to return the next item in the sequence to return or to raise the `StopIteration` exception. This is used to indicate that the iterator has finished supplying values.
            
- `__iter__()` from the iterable protocol which is used to return the iterator object

- `__next__()` from the iterator protocol which is used to obtain the next value in a sequence of values
            
              

```python
class Fibonacci:
    pass

f = Fibonacci()
#won't work since there is no _iter__ method
#for x in f:
    #print(x)
```

```python
#corrected version
class Fibonacci:
    def __iter__(self):
        yield 5
        yield 7
        yield 11

f = Fibonacci()
for x in f:
    print(x)
```

```python
#other version

class Fibonacci:
    def __olditer__(self):
        yield 5
        yield 7
        yield 11
    def __iter__(self):
        self.p = 0
        self.q = 1
        return self
    def __next__(self):
        r = self.p + self.q
        if r > 1000:
            raise StopIteration() #sets condition under which __next__ based loop stops
        self.p = self.q
        self.q = r
        return r

f = Fibonacci()

for x in f:
    print(x)

print("==========")
i = f.__iter__()
while True:
    try:
        x = i.__next__()
    except StopIteration:
        break
    print(x)
```

```python
import threading

def oddNumbers():
    k = 1
    while k <= 1000:
        mutex.acquire()
        print(k)
        mutex.release()
        k += 2

def evenNumbers():
    k = 2
    while k <= 1000:
        mutex.acquire()
        print(k)
        mutex.release()
        k += 2


mutex = threading.Lock() # creates a mutex object

# Create threads
thread1 = threading.Thread(target=oddNumbers)
thread2 = threading.Thread(target=evenNumbers)

# Start threads
thread1.start()
thread2.start()

# Wait for threads to complete
thread1.join()
thread2.join()
```

```python
import http.server,urllib

class MyWebApp(http.server.BaseHTTPRequestHandler):
    def do_GET(self):

        h = urllib.parse.urlsplit(self.path) # parses the path of the request, which includes the query string, and returns a tuple containing the different components of the URL
        #print(self.path) # prints the path of the request, which includes the query string
        #print(h.path) # prints the path component of the parsed URL
        #print(h.query) # prints the query string component of the parsed URL

        r = urllib.parse.parse_qsl(h.query) # parses the query string into a list of key-value pairs, where each pair is a tuple containing the key and value as strings
        #print(r)

        params = dict(r) # converts the list of key-value pairs into a dictionary, where the keys are the parameter names and the values are the parameter values
        #print(params)
        
        if 'a' in params and 'b' in params:
            a = int(params['a']) # retrieves the value of parameter 'a' from the dictionary and converts it to an integer
            b = int(params['b']) # retrieves the value of parameter 'b' from the dictionary and converts it to an integer
            c = a + b # calculates the sum of a and b
        else:
            c = 0
        
        resp = "<html><body>Sum: " + str(c) + "</body></html>" # creates a response message containing the result of the calculation
        
        byteresp = bytes(resp, "utf-8")
        self.send_response(200)
        self.send_header("content-type", "text/html; charset=utf-8")
        self.send_header("content-length", str(len(byteresp)) )
        self.end_headers()
        self.wfile.write(byteresp)



srv = http.server.ThreadingHTTPServer( ("0.0.0.0", 8000), MyWebApp  )
srv.serve_forever()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# RGB inputs (normalized 0–1)
X = torch.tensor([
    [1, 1, 1],   # White
    [0, 0, 0],   # Black
    [1, 1, 0],   # Yellow
    [1, 0, 1],   # Magenta
    [1, 0.4, 1],   # Magenta
    [0, 1, 1],   # Cyan
    [0, 1, 0],   # Green
    [1, 0, 0],   # Red
    [0, 0, 1],    # Blue
    [0.75, 0.75, 0.75], #Silver
], dtype=torch.float32)

# Labels (class indices)
y = torch.tensor([0,1,2,3,3,4,5,6,7,8])

# Simple neural network
model = nn.Sequential(
    nn.Linear(3, 16),
    nn.ReLU(),
    nn.Linear(16, 9)
)

# Loss + optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# Training loop
for epoch in range(500):
    outputs = model(X)
    loss = criterion(outputs, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Test function
colors = ["White","Black","Yellow","Magenta","Cyan","Green","Red","Blue", "Silver"]

def predict(rgb):
    with torch.no_grad():
        input_tensor = torch.tensor([rgb], dtype=torch.float32)
        output = model(input_tensor)
        predicted = torch.argmax(output, dim=1).item()
        return colors[predicted]

# Test examples
print(predict([0.75, 0.75, 0.1]))   # Yellow
print(predict([0.25, 0.75, 0.1]))   # Green
print(predict([0.91, 0.89, 0.93]))   # White
print(predict([0.95, 0.5, 0.97]))   # Magenta
print(predict([0.75, 0.75, 0.75]))
```
