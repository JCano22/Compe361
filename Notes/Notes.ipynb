### ***While loops***
-Similar to C++: 

`while <condition>:
    do something`

```python
# Greatest common divisor

p = 75
q = 45

# while they are not equal, will subtract smallest from largest until numbers are equal indicating GCD

while p != q:
    if p > q:
        p -= q
    else:
        q -= p

print(p)
```

-For loops and While loops can have an `else: ` section following the loop.

-The `else: ` block only runs after the while loop has been exited normally and not by `break`

```python
# Checking for prime numbers

p = 997
j = 2

# while j is less than p, it will check if it is divisible by j, if it is 'Not prime' and break the loop, otherwise it will continue to check unitl they are equal, and then the `else:` clause activates to print 'Prime'

while j < p:
    if p % j == 0:
        print('Not prime.')
        break
    j += 1
else:
    print('Prime.')
```

### ***decimal*** library

-The Python decimal module provides support for correctly rounded decimal floating-point arithmetic. It offers several advantages over the built-in floating-point type, including preserving significant digits.

```python
from decimal import Decimal

p = 0.1 #0.0999999
q = 0.2 #0.1999999
r = p + q #0.2999999
print(r)

a = Decimal('0.1') #passing a string so it gets the exact value we intend, instead of inheriting the inaccuracies of binary floating-point numbers.
b = Decimal('0.2')
c = a + b
print(c)
```

#### ***Quantize** function*

-quantize() method is used to round or format a decimal number to a specific number of decimal places while following a chosen rounding rule.

```python
from decimal import Decimal, ROUND_HALF_UP

a = Decimal('0.17')
b = Decimal('0.28')
c = a + b
print(c)
r = c.quantize(Decimal("0.0"))
print(r)

r = c.quantize(Decimal("0.0"), ROUND_HALF_UP)
print(r)

r = c.quantize(Decimal("0.01"), ROUND_HALF_UP)
print(r)
```

**GCD function** example with `return`:

```python
def GCD(p, q):
    while p != q:
        if p > q:
            p = p - q
        else:
            q = q - p
    return p

u = 75
z = 45

r = GCD(u, z)

print(r)
print(u)
print(z)
```

**LCM function** example:

-Least common multiple is the *smallest* number that can be divided by both numbers.

- We use the double slash: `//` for integer division

```python
def GCD(p, q):
    while p != q:
        if p > q:
            p = p - q
        else:
            q = q - p
    return p


def LCM(a, b):
    return a * b // GCD(a,b)

u = 75
z = 45

r = LCM(u,z)

print(r)
print(u)
print(z)
```

**Example of function** parameters vs arguments:

```python
def Exchange(p, q):
    r = p
    p = q
    q = r

u = 75
z = 89

Exchange(u, z)
print(u)
print(z)

# when the function is called, a copy of the variables is sent to the function.  Even though the variables are pointers to the address, they don't send the address to the function.  Therefore the arguments aren't modified in the function.
```

For the above example:

- In Python, variables are names bound to objects, not memory locations or pointers like in C.

- When `Exchange(u, z)` is called:

    - `p` is bound to the same object as `u`

    - `q` is bound to the same object as `z`

- The function receives **copies of the references**, not the variable names `u` and `z` themselves.

- Inside the function, statements like `p = q` and `q = r`:

    - Rebind the local names `p` and `q`

    - Do not affect the bindings of `u` and `z` in the caller

- Integers are **immutable**, so their values cannot be modified in place.

- As a result, swapping `p` and `q` only changes local variables inside the function.

- Therefore, `u` and `z` remain unchanged after the function call.

### ***Local and Global* variables**

-variables created in a function are ***local variables***


```python
def sum(x, z):
    u = x * x + z * z #u is a local variable
    return u

def multiply(x, z):
    u = x * x * z * z #u is also a local variable here
    return u 

# both u variables in the previous functions are separate and don't intersect with each other

def Echo():
    print(u) # since there is no local variable `u` it will look for global variable.

u = 2500 #this is a global variable

r = sum(2, 3)
print(r)
Echo()
```

### *self*

- `self` is a special variable in Python that refers to the current object. It is the first parameter of every instance method. When you call a method, you do not pass `self` explicitly, Python automatically provides it.

- `self` represents the object in which the method is running, giving the method access to that object’s data (attributes) and other methods. In short, `self` is the object itself.

```python
class Person:
    def __init__(self, name, age):
        print('init called with', name, age)
        self.name = name
        self.age = age
    def echo(self):
        print(self.name)
        print(self.age)

p = Person('Jorge', 40)
p.echo()
```

### class constructors

- A constructor is a special method used to initialize a new object

- In Python, the constructor is named `__init__`

- It runs automatically when an object is created

- `__init__` must include self as the first parameter

- Additional parameters are passed during object creation

- Instance variables are created inside `__init__` using self

- Without self, variables are local and discarded

### Rectangle *class*


```python
class Rectangle:
    length = None
    width = None

    def __init__(self, l, w):
        self.length = l
        self.width = w
    
    def area(self):
        return (self.length * self.width)
    
    def perimeter(self):
        return (2 * self.length + 2 * self.width)
    
r = Rectangle(2, 4)
a = r.area()
p = r.perimeter()

print(r)
print(r.length)
print(a)
print(p)
```

- You can make another class be a 'child' class of another. 
    - Here `Cuboid` is a child class of `Rectangle`.
    - It inherits the methods from the parent class
    - You can call the parent methods for the `Cuboid` objects using `super()`

Example:

```python
# parent class
class Rectangle:
    length = None
    width = None

    def __init__(self, l, w):
        self.length = l
        self.width = w
    
    def area(self):
        return (self.length * self.width)
    
    def perimeter(self):
        return (2 * self.length + 2 * self.width)
    
# child class
class Cuboid(Rectangle):
    pass

r = Rectangle(2, 4)
a = r.area()
p = r.perimeter()

print(r.length)
print(a)
print(p)

c = Cuboid(9, 3)
b = c.area()
print(f'Cuboid area: {b}')
```

- You can add its own methods as well

- In code below, we overwrite the `__str__` method inside the parent class.  We call to print cuboid object, and it calls parent class `__str__`. But it calls child class method

```python
# parent class
class Rectangle:
    length = None
    width = None

    def __init__(self, l, w):
        self.length = l
        self.width = w
    
    def area(self):
        return (self.length * self.width)
    
    def perimeter(self):
        return (2 * self.length + 2 * self.width)
    
    #overwriting __str__ method
    def __str__(self):
        a = self.area()
        return "The area is: " + str(a)
    
# child class
class Cuboid(Rectangle):
    height = None

    def __init__(self, l, w, h):
        self.height = h
        #you can call the parent's constructor for the remaining values (l, w) using `super` followed by the method you wish to call
        super().__init__(l, w)
    
    def area(self):
        return 2 * super().area() + 2 * (self.length + self.width) * self.height


r = Rectangle(2, 4)
a = r.area()
p = r.perimeter()

print(r.length)
print(a)
print(p)

c = Cuboid(9, 5, 2)
b = c.area()

print(f'Cuboid area: {b}')

print(c) #currently prints address as there is no __str__ method
```
